```python
from transitions.extensions import HierarchicalGraphMachine as HGM
from transitions.extensions.nesting import NestedState as NS
```

```python
class Matter(object):
    pass
lump = Matter()
machine = HGM(model=lump, initial="new")
```

```python
machine.add_states(["filled", "summited", "approved", "executing", "done"])
```

```python
machine.add_transition("filled_table", "new", "filled")
```

```python
machine.add_transition('summit', "filled", "summited")
```

```python
machine.add_transition("approve", "summited", "approved")
```

```python
machine.add_transition("execute", "approved", "executing")
```

```python
machine.add_transition("confirm", "executing", "done")
```
